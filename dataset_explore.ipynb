1️⃣ Load and Inspect the Dataset

```python
import pandas as pd

# Load dataset
df = pd.read_csv("data/StudentPerformanceFactors.csv")

# Basic inspection
print(df.shape)
print(df.head())
print(df.info())
```

2️⃣Handle Missing Values

```python
# Separate numerical and categorical columns
num_cols = df.select_dtypes(include=["int64", "float64"]).columns
cat_cols = df.select_dtypes(include=["object"]).columns

# Fill missing numerical values with median
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values with mode
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

3️⃣ Drop Non-Informative / Identifier Columns

```python
id_like_cols = [col for col in df.columns if "id" in col.lower()]

df = df.drop(columns=id_like_cols)
```

4️⃣ Encode Categorical Features

```python
df_encoded = pd.get_dummies(df, columns=cat_cols, drop_first=True)
```

5️⃣ Feature Selection for Autoencoder

```python
label_like_cols = ["FinalGrade", "PassFail", "ExamResult"]

features = [col for col in df_encoded.columns if col not in label_like_cols]
X = df_encoded[features]
```

6️⃣ Normalize the Data (Critical Step)

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

7️⃣ Final Sanity Check

```python
import numpy as np

print("Shape:", X_scaled.shape)
print("Mean (approx):", np.mean(X_scaled))
print("Std (approx):", np.std(X_scaled))
```

✅ Final Output (What You Now Have)

You now have:

- X_scaled → clean, normalized input for the autoencoder

- No missing values

- No categorical leakage

- No identifiers

- Reproducible preprocessing pipeline
